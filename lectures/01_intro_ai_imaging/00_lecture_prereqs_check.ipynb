## Setup helpers

Run this cell first. It defines the helper functions every check below uses.


```python
import importlib
import importlib.metadata
import platform
import sys

def show_runtime():
    print(f"Platform : {platform.platform()}")
    print(f"Python   : {sys.version.split()[0]}  ({sys.executable})")

def check_module(module_name, why, fix, required=True, min_version=None):
    """Try to import a module; report status, version, and fix if missing."""
    try:
        mod = importlib.import_module(module_name)
        version = getattr(mod, "__version__", None)
        if version is None:
            try:
                version = importlib.metadata.version(module_name)
            except importlib.metadata.PackageNotFoundError:
                version = "unknown"
        if min_version:
            try:
                cur = tuple(int(x) for x in str(version).split(".")[:2])
                mn  = tuple(int(x) for x in min_version.split(".")[:2])
                if cur < mn:
                    print(f"⚠  {module_name:<22} version {version} < required {min_version}")
                    print(f"   Why : {why}")
                    print(f"   Fix : {fix}")
                    return False
            except ValueError:
                pass
        print(f"✓  {module_name:<22} {version}")
        return True
    except ImportError as e:
        symbol = "✗" if required else "○"
        marker = "REQUIRED" if required else "optional"
        print(f"{symbol}  {module_name:<22} NOT INSTALLED ({marker})")
        print(f"   Why : {why}")
        print(f"   Fix : {fix}")
        return False

show_runtime()
```
